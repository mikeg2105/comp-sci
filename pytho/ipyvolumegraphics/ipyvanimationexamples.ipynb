```python
# create 2d grids: x, y, and r
u = np.linspace(-10, 10, 25)
x, y = np.meshgrid(u, u)
r = np.sqrt(x**2+y**2)

v = np.linspace(-np.pi, np.pi, 25)
xv,yv=np.meshgrid(np.cos(v),np.sin(v))
print("x,y and z are of shape", x.shape)
# and turn them into 1d
x = x.flatten()
y = y.flatten()
r = r.flatten()
print("and flattened of shape", x.shape)
```

```python
# create a sequence of 15 time elements
time = np.linspace(0, np.pi*2, 15)
z = np.array([(np.cos(r + t) * np.exp(-r/5)) for t in time])
print("z is of shape", z.shape)
```

```python
# draw the scatter plot, and add controls with animate_glyphs
ipv.figure()
s = ipv.scatter(x, z, y, marker="sphere")
ipv.animation_control(s, interval=200)
ipv.ylim(-3,3)
ipv.show()
```

```python
# Now also include, color, which containts rgb values
color = np.array([[np.cos(r + t), 1-np.abs(z[i]), 0.1+z[i]*0] for i, t in enumerate(time)])
size = (z+1)
print("color is of shape", color.shape)
```

```python
color = np.transpose(color, (0, 2, 1)) # flip the last axes
```

```python
ipv.figure()
s = ipv.scatter(x, z, y, color=color, size=size, marker="arrow")
ipv.animation_control(s, interval=200)
ipv.ylim(-3,3)
ipv.show()
```

```python
import numpy as np
import ipyvolume as ipv

# Parameters
nx, ny = 64, 64              # Grid resolution
nt = 60                      # Number of time steps
x = np.linspace(-2, 2, nx)
y = np.linspace(-2, 2, ny)
X, Y = np.meshgrid(x, y)

# Generate time-evolving scalar field
data = []
for t in np.linspace(0, 2*np.pi, nt):
    # Magnitude of a 2D Gaussian modulated by sin(t)
    Z = np.exp(-X**2 - Y**2) * (0.5 + 0.5*np.sin(t))
    data.append(Z)

# Convert to 3D array: shape (nt, nx, ny)
volume = np.array(data)

# Normalize for better display
volume = volume / np.max(volume)

# Plot
ipv.figure()
s=ipv.volshow(volume, level=0.3, opacity=0.4, data_min=0, data_max=1)
ipv.animation_control(s)  # Add time slider
ipv.show()
```

```python
import numpy as np
import ipyvolume as ipv
import ipywidgets as widgets






# Grid setup
nx, ny = 64, 64
nt = 60
x = np.linspace(-2, 2, nx)
y = np.linspace(-2, 2, ny)
X, Y = np.meshgrid(x, y)

# Generate time-dependent frames
frames = []
for t in np.linspace(0, 2*np.pi, nt):
    Z = np.exp(-X**2 - Y**2) * (0.5 + 0.5*np.sin(t))
    frames.append(Z)
frames = np.array(frames)

# Normalize
frames /= np.max(frames)

# Initial visualization
ipv.figure()
vol = ipv.volshow(frames[0][np.newaxis, :, :], level=0.3, opacity=0.4)
#vol = ipv.volshow(frames[0][np.newaxis, :, :], level=0.3, opacity=0.4)







# Volume needs depth > 1 for lighting
def make_3d_slice(scalar_2d):
    return np.stack([0.9*scalar_2d, scalar_2d, 1.1*scalar_2d])

# Initial 3D slice
ipv.figure()
vol = ipv.volshow(make_3d_slice(frames[0]), level=0.3, opacity=0.4)

# Slider logic
def update_frame(frame_idx):
    vol.data = make_3d_slice(frames[frame_idx])
    return vol

slider = widgets.IntSlider(min=0, max=nt-1, step=1, value=0, description='Time')
widgets.interact(update_frame, frame_idx=slider)
ipv.show()




```

```python
import numpy as np
import ipyvolume as ipv
import ipywidgets as widgets

# Grid size
nx, ny, nz = 64, 64, 3
nt = 60

# Coordinate mesh
x = np.linspace(-2, 2, nx)
y = np.linspace(-2, 2, ny)
X, Y = np.meshgrid(x, y)

# Function to create a padded 3D scalar volume
def make_3d_volume(Z2d):
    return np.stack([
        0.95 * Z2d,
        Z2d,
        1.05 * Z2d
    ], axis=-1)  # Final shape: (nx, ny, nz)

# Generate frames
frames = []
for t in np.linspace(0, 2*np.pi, nt):
    Z2d = np.exp(-X**2 - Y**2) * (0.5 + 0.5*np.sin(t))
    V3d = make_3d_volume(Z2d)
    frames.append(V3d)

frames = np.array(frames)  # Shape: (nt, nx, ny, nz)
frames /= np.max(frames)

# Set up visualization
ipv.figure()
volume = ipv.volshow(frames[0], level=0.3, opacity=0.4)

# Interactive slider
def update_volume(t_idx):
    volume.data = frames[t_idx]
    return volume

slider = widgets.IntSlider(min=0, max=nt-1, step=1, value=0, description="Time")
widgets.interact(update_volume, t_idx=slider)
ipv.show()
```
